```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
```

```python
df = pd.read_csv('../Dataset/Mall_Customers.csv', encoding='latin-1')
print(df.head())
```

```python
df = df[["Annual Income (k$)", "Spending Score (1-100)"]]
df.columns = ["Annaul Income", "Spending Score"]
print(df.head())
```

```python
X = df[["Annaul Income", "Spending Score"]]

scaler = StandardScaler()
X_scaler = scaler.fit_transform(X)
```

```python
model = KMeans(
    n_clusters=2,
    init="k-means++",
    n_init=10,
    random_state=42
)

model.fit(X_scaler)
```

## For each value which cluster is assigned

```python
df["Cluster"] = model.labels_

print(df)
```

## Where are the centroids ?

```python
print(model.cluster_centers_)
```

## Predicting new customer

```python
new_customer = pd.DataFrame({
    "Annaul Income": [20],
    "Spending Score": [8]
})

new_scaled = scaler.transform(new_customer)

cluster_predict = model.predict(new_scaled)
print(cluster_predict)
```

## Visualization

```python
plt.scatter(df["Annaul Income"], df["Spending Score"], c=df["Cluster"], cmap="viridis")
plt.xlabel("Annual Income (k$)")
plt.ylabel("Spending Score")
plt.title("Customer Segments")
plt.show()
```

## Import Libraries

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
```

## Dataset Preparation

```python
df = pd.read_csv('../Dataset/Mall_Customers.csv', encoding='latin-1')
df = df[["Annual Income (k$)", "Spending Score (1-100)"]]
df.columns = ["Annual Income", "Spending Score"]

X = df[["Annual Income", "Spending Score"]]
print(X.head())
```

## Scale the features

```python
scaler = StandardScaler()
X_scaler = scaler.fit_transform(X)
```

## Compare different values of K

```python
results = []

for k in range(1, 7):
    model = KMeans (
        n_clusters=k,
        init="k-means++",
        n_init=10,
        random_state=42
    )

    labels = model.fit_predict(X_scaler)

    row = {
        "K": k,
        "Interia": model.inertia_
    }

    if k >= 2:
        row["Silhouette"] = silhouette_score(X_scaler, labels)

    results.append(row)

result_df = pd.DataFrame(results)
print(result_df)

```

## Train your selected model

```python
final_model = KMeans (
    n_clusters=5,
    init="k-means++",
    n_init=10,
    random_state=42
)

df['Clusters'] = final_model.fit_predict(X_scaler)
```

## Interpret the clusters

```python
cluster_summary = df.groupby("Clusters").agg(
    Annual_Count=("Annual Income", "count"),
    Average_Annual=("Annual Income", "mean"),
    Average_Spending=("Spending Score", "mean")
)

print(cluster_summary)
```

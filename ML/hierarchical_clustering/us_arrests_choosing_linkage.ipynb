## Libraries

```python
import pandas as pd
from sklearn.cluster import AgglomerativeClustering
from sklearn.metrics import silhouette_score
from sklearn.preprocessing import StandardScaler
```

## Dataset Preparation

```python
import statsmodels.api as sm
df = sm.datasets.get_rdataset("USArrests", "datasets").data
print(df.head())
```

## Standardize

```python
X = df[["Murder", "Assault", "UrbanPop", "Rape"]]
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

## Compare the Linkage methods

```python
result = []

for method in ["complete", "average", "ward", "single"]:
    model = AgglomerativeClustering(
        n_clusters=5,
        linkage=method,
        metric="euclidean"
    )

    labels = model.fit_predict(X_scaled)

    result.append({
        "Linkage": method,
        "Silhouette_sample": silhouette_score(X_scaled, labels),
        "Smallest_cluster": pd.Series(labels).value_counts().min(),
        "Largest_cluster": pd.Series(labels).value_counts().max()
    })
```

```python
results_df = pd.DataFrame(result)
print(results_df.sort_values("Silhouette_sample", ascending=False))
```

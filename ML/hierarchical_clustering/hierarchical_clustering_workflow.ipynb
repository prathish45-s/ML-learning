## Libraries

```python
import pandas as pd
from sklearn.cluster import AgglomerativeClustering
from sklearn.metrics import silhouette_score
from sklearn.preprocessing import StandardScaler
```

## Dataset Preparation

```python
import statsmodels.api as sm
df = sm.datasets.get_rdataset("USArrests", "datasets").data
print(df.head())
```

## Standardize

```python
X = df[["Murder", "Assault", "UrbanPop", "Rape"]]
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

## Creating cluster

```python
model = AgglomerativeClustering(
    n_clusters=5,
    linkage="ward"
)

labels = model.fit_predict(X_scaled)
```

## Attaching the labels to the original data

```python
df["Cluster"] = labels
```

## Interpret clusters in original units

```python
print(
    df.groupby("Cluster").agg(
        Arrest_Count=("Cluster", "size"),
        Average_Murder=("Murder", "mean"),
        Average_Rape=("Rape", "mean")
    )
)
```

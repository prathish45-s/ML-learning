## Library

```python
import numpy as np
import pandas as pd
from sklearn.cluster import DBSCAN
```

## Dataset

```python
X = np.array([
    [1.0], [1.1], [1.2], [2.0],
    [5.0], [5.1], [5.2], [9.0]
])
```

## Min_sample and Eps confige

```python
settings = [
    (0.4, 3),
    (0.8, 3),
    (3.0, 3),
    (0.4, 4)
]
```

## Multi setting in model

```python
result = []
for eps, min_sample in settings:
    model = DBSCAN (
        eps=eps,
        min_samples=min_sample
    )
    labels = model.fit_predict(X)
    n_cluster = len(set(labels) - {-1})
    n_noise = np.sum(labels == -1)

    result.append ({
         "eps": eps,
        "min_samples": min_sample,
        "clusters": n_cluster,
        "noise_points": n_noise
    })
```

## Result

```python
print(pd.DataFrame(result))
```

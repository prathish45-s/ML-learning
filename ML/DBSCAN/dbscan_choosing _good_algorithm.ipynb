## Libraries


```python
import numpy as np
import pandas as pd
from sklearn.cluster import DBSCAN
from sklearn.datasets import make_blobs
import time
```

## Import dataset

```python
X, _ = make_blobs(
n_samples=750, centers=4, cluster_std=0.5, random_state=0, n_features=100
)
```

## Algorithm Selection

```python
algorithm = ["kd_tree", "ball_tree", "brute", "auto"]

for algo in algorithm:
    model = DBSCAN(
        min_samples=5,
        eps=8,
        algorithm=algo,
        leaf_size=30
    )

    start = time.perf_counter()
    model.fit(X)
    elapsed = time.perf_counter() - start

    lables = model.labels_
    n_cluster = len(set(lables)) - (1 if -1 in lables else 0)
    n_noise = np.sum(lables == -1)

    print(
        f"{algo:10s} | "
        f"Time taken: {elapsed:.4f} seconds | "
        f"Number of clusters: {n_cluster} | "
        f"Number of noise points: {n_noise}"
    )
```

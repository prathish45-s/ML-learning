## Library

```python
from sklearn.cluster import DBSCAN
import numpy as np
```

## Dataset

```python
X = [
    [1.0], [1.1], [1.2],
    [2.0],
    [5.0], [5.1], [5.2],
    [9.0]
]
```

## Training

```python
model = DBSCAN(
    eps=0.4,
    min_samples=3
)
labels = model.fit_predict(X)
```

## Output

```python
print(labels)
```

## Count the clusters and noise points

```python
n_cluster = len(set(labels)) - (1 if -1 in labels else 0)
n_noise = np.sum(labels == -1)
```

```python
print("No.of clusters: ", n_cluster)
print("No. of noise: ", n_noise)
```

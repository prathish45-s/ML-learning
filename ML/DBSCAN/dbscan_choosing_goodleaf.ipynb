## Import Libraries

```python
from sklearn.cluster import DBSCAN
from sklearn.datasets import make_blobs
import pandas as pd
import numpy as np
import time
```

## Dataset

```python
X, _ = make_blobs(
    n_samples=5000,
    n_features=2,
    centers=4,
    random_state=42
)

leaf_size = [10, 20, 30, 60, 100]
```

## Choosing best Leaf size

```python
for leaf in leaf_size:
    times = []

    for i in range(3):
        model = DBSCAN(
            eps=0.6,
            min_samples=5,
            algorithm="kd_tree",
            leaf_size=leaf
        )

        start = time.perf_counter()
        model.fit(X)
        times.append(time.perf_counter() - start)
    labels = model.labels_
    clusters = len(set(labels)) - (1 if -1 in labels else 0)
    noise = np.sum(labels == -1)

    print(
        f"leaf_size={leaf:3d} | "
        f"time={np.median(times):.5f}s | "
        f"clusters={clusters} | "
        f"noise={noise}"
    )
```
